```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../Dataset/HR Analytics.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

df.info()
```

```python
# Check dataset dimensions
print("Rows and Columns:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset Information:")
df.info()
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
# Total employees
total_employees = len(df)

# Employees who left the company
employees_left = (df["Attrition"] == "Yes").sum()

# Employees who stayed
employees_stayed = (df["Attrition"] == "No").sum()

# Attrition rate
attrition_rate = (employees_left / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Left:", employees_left)
print("Employees Stayed:", employees_stayed)
print("Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
department_analysis = df.groupby("Department").agg(
    Total_Employees=("EmployeeNumber", "count"),
    Employees_Left=("Attrition", lambda x: (x == "Yes").sum())
)

department_analysis["Attrition_Rate"] = (
    department_analysis["Employees_Left"]
    / department_analysis["Total_Employees"]
    * 100
).round(2)

print(department_analysis)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

department_attrition = df.groupby('Department').apply(
    lambda x: (x['Attrition'] == 'Yes').mean() * 100,
    include_groups=False
)

department_attrition = department_attrition.sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=department_attrition.index,
    y=department_attrition.values
)

plt.title('Department-wise Employee Attrition Rate')
plt.xlabel('Department')
plt.ylabel('Attrition Rate (%)')

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

overtime_attrition = df.groupby('OverTime').apply(
    lambda x: (x['Attrition'] == 'Yes').mean() * 100,
    include_groups=False
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=overtime_attrition.index,
    y=overtime_attrition.values
)

plt.title('Overtime vs Employee Attrition Rate')
plt.xlabel('Overtime')
plt.ylabel('Attrition Rate (%)')

plt.tight_layout()
plt.show()
```

```python
job_role_attrition = df.groupby('JobRole').apply(
    lambda x: (x['Attrition'] == 'Yes').mean() * 100,
    include_groups=False
)

job_role_attrition = job_role_attrition.sort_values(
    ascending=False
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=job_role_attrition.values,
    y=job_role_attrition.index
)

plt.title('Job Role vs Employee Attrition Rate')
plt.xlabel('Attrition Rate (%)')
plt.ylabel('Job Role')

plt.tight_layout()
plt.show()
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 24, 34, 44, 54, 100],
    labels=['Under 25', '25-34', '35-44', '45-54', '55+']
)

age_attrition = df.groupby('Age_Group', observed=False).apply(
    lambda x: (x['Attrition'] == 'Yes').mean() * 100,
    include_groups=False
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=age_attrition.index,
    y=age_attrition.values
)

plt.title('Age Group vs Employee Attrition Rate')
plt.xlabel('Age Group')
plt.ylabel('Attrition Rate (%)')

plt.tight_layout()
plt.show()
```

```python
income_attrition = df.groupby('Attrition')['MonthlyIncome'].mean()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=income_attrition.index,
    y=income_attrition.values
)

plt.title('Average Monthly Income vs Employee Attrition')
plt.xlabel('Employee Attrition')
plt.ylabel('Average Monthly Income')

plt.tight_layout()
plt.show()
```

```python
satisfaction_attrition = df.groupby('JobSatisfaction').apply(
    lambda x: (x['Attrition'] == 'Yes').mean() * 100,
    include_groups=False
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=satisfaction_attrition.index,
    y=satisfaction_attrition.values
)

plt.title('Job Satisfaction vs Employee Attrition Rate')
plt.xlabel('Job Satisfaction Score (1-4)')
plt.ylabel('Attrition Rate (%)')

plt.tight_layout()
plt.show()
```
